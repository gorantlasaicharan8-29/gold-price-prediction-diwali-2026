```python
import importlib

required_libraries = {
    "pandas": "pandas",
    "numpy": "numpy",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "sklearn": "scikit-learn",
    "yfinance": "yfinance",
    "xgboost": "xgboost",
    "statsmodels": "statsmodels",
}

all_libraries_loaded = True

for module_name, display_name in required_libraries.items():
    try:
        module = importlib.import_module(module_name)
        version = getattr(module, "__version__", "version unavailable")
        print(f"SUCCESS: {display_name} loaded successfully (version: {version})")
    except ImportError as error:
        all_libraries_loaded = False
        print(f"FAILED: {display_name} could not be loaded ({error})")

if all_libraries_loaded:
    print("SUCCESS: All required ML and data-analysis libraries are available.")
else:
    print("WARNING: One or more required libraries are unavailable.")
```
